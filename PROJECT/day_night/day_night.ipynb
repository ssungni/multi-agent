```python
import os
from dotenv import load_dotenv

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
# GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")
ANTHROPIC_API_KEY = os.getenv("ANTHROPIC_API_KEY")

# print(OPENAI_API_KEY)
# print(GOOGLE_API_KEY)
# print(ANTHROPIC_API_KEY)
```

```python
from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_anthropic import ChatAnthropic

gpt = ChatOpenAI(
    model="gpt-4.1-mini",
    api_key=OPENAI_API_KEY
)

# gemini = ChatGoogleGenerativeAI(
#     model="gemini-3.5-flash",
#     google_api_key=GOOGLE_API_KEY
# )

claude = ChatAnthropic(
    model="claude-sonnet-5",
    api_key=ANTHROPIC_API_KEY
)
```

```python
question = "'오늘 날씨가 좋네요.'를 영어로 번역해 주세요."

models = {
    "GPT": gpt,
    # "Gemini": gemini,
    "Claude": claude
}

for name, model in models.items():
    print(name)

    try:
        response = model.invoke(question)

        if response and response.content:
            print(response.content)
        else:
            print("Error: Response received, but no content was returned.")

    except Exception as e:
        print(f"Error: API request failed: {e}")

    print("*" * 50)
```

```python
from openai import OpenAI
import base64

client = OpenAI(api_key=OPENAI_API_KEY)

with open(r"C:\Users\User\OneDrive\Desktop\day_night\day.jpg", "rb") as image_file:
    result = client.images.edit(
        model="gpt-image-2",
        image=image_file,
        prompt=(
            "Convert this daytime photo into a realistic nighttime scene. "
            "Keep the buildings, roads, trees, and composition exactly the same. "
            "Only change the lighting and atmosphere to night. "
            "Add moonlight, streetlights, and naturally illuminated windows."
        ),
        size="1024x1024",
        quality="high",
    )

with open("night.jpg", "wb") as f:
    f.write(base64.b64decode(result.data[0].b64_json))

print("night.jpg 저장 완료!")
```
